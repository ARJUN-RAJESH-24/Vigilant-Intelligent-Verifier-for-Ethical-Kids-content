```python
# 01_preprocessing.ipynb

import pandas as pd
import os

# Create a sample captions dataset
captions = pd.DataFrame({
    "id": ["vid1", "vid2", "vid3", "vid4"],
    "caption": [
        "This is a beach party video!",
        "Romantic couple kiss scene 💋",
        "Cooking a healthy dinner at home",
        "Hot dance performance in club"
    ]
})
os.makedirs("../data", exist_ok=True)
captions.to_csv("../data/captions.csv", index=False)

# Create a label dataset (1 = adult, 0 = safe)
labels = pd.DataFrame({
    "id": ["vid1", "vid2", "vid3", "vid4"],
    "label": [0, 1, 0, 1]
})
labels.to_csv("../data/labels.csv", index=False)

print("✅ Sample captions.csv and labels.csv created in data/")

# View
pd.read_csv("../data/captions.csv").head()
```
