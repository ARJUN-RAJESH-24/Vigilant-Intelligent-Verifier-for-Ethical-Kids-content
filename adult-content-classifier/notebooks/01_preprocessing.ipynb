```python
# 01_preprocessing.ipynb - Data Setup
# This notebook creates sample datasets for the adult content classifier

import pandas as pd
import os
import numpy as np

print("="*80)
print("📁 PREPROCESSING - CREATING SAMPLE DATASET")
print("="*80)

# Create data directory
os.makedirs("../data", exist_ok=True)
os.makedirs("../data/videos", exist_ok=True)

# ============================================================================
# 1. CREATE SAMPLE CAPTIONS
# ============================================================================
print("\n1️⃣ Creating sample captions...")

captions = pd.DataFrame({
    "id": [
        "vid001", "vid002", "vid003", "vid004", "vid005",
        "vid006", "vid007", "vid008", "vid009", "vid010",
        "vid011", "vid012", "vid013", "vid014", "vid015"
    ],
    "caption": [
        # Safe content
        "Beautiful beach sunset with family enjoying vacation time",
        "Cooking a healthy dinner at home with fresh vegetables",
        "Morning yoga routine for beginners - peaceful meditation",
        "Kids playing in the park on a sunny afternoon",
        "Travel vlog exploring ancient temples in Thailand",
        
        # Adult content indicators
        "Hot dance performance at nightclub with DJ",
        "Romantic couple kiss scene from movie trailer",
        "Intimate bedroom scene with soft lighting",
        "Sexy model photoshoot behind the scenes footage",
        "Late night party with provocative dancing",
        
        # More safe content
        "Educational science experiment for students",
        "DIY home renovation project step by step",
        "Cute puppies playing in the garden compilation",
        "Professional makeup tutorial for beginners",
        "Gaming walkthrough final boss battle strategy"
    ]
})

captions.to_csv("../data/captions.csv", index=False)
print(f"   ✅ Created captions.csv with {len(captions)} entries")

# ============================================================================
# 2. CREATE LABELS
# ============================================================================
print("\n2️⃣ Creating sample labels...")

labels = pd.DataFrame({
    "id": [
        "vid001", "vid002", "vid003", "vid004", "vid005",
        "vid006", "vid007", "vid008", "vid009", "vid010",
        "vid011", "vid012", "vid013", "vid014", "vid015"
    ],
    "label": [
        # 0 = Safe, 1 = Adult
        0, 0, 0, 0, 0,  # Safe content
        1, 1, 1, 1, 1,  # Adult content
        0, 0, 0, 0, 0   # Safe content
    ]
})

labels.to_csv("../data/labels.csv", index=False)
print(f"   ✅ Created labels.csv with {len(labels)} entries")

# ============================================================================
# 3. DISPLAY STATISTICS
# ============================================================================
print("\n3️⃣ Dataset Statistics:")
print(f"   Total samples: {len(labels)}")
print(f"   Safe (0): {(labels['label'] == 0).sum()} ({(labels['label'] == 0).sum()/len(labels)*100:.1f}%)")
print(f"   Adult (1): {(labels['label'] == 1).sum()} ({(labels['label'] == 1).sum()/len(labels)*100:.1f}%)")

# ============================================================================
# 4. PREVIEW DATA
# ============================================================================
print("\n4️⃣ Preview of Created Data:")
print("\n📝 Captions Sample:")
print(captions.head(10))

print("\n🏷️  Labels Sample:")
print(labels.head(10))

# ============================================================================
# 5. MERGE AND VALIDATE
# ============================================================================
print("\n5️⃣ Validating Data Integrity...")

merged = captions.merge(labels, on='id', how='inner')
print(f"   ✅ Successfully merged: {len(merged)} records")

# Check for missing values
print(f"   Missing captions: {captions['caption'].isna().sum()}")
print(f"   Missing labels: {labels['label'].isna().sum()}")

# Display merged sample
print("\n📊 Merged Data Sample (Caption + Label):")
sample = merged.head(5)
for idx, row in sample.iterrows():
    label_text = "🔴 ADULT" if row['label'] == 1 else "🟢 SAFE"
    print(f"\n{row['id']}: {label_text}")
    print(f"   Caption: {row['caption'][:60]}...")

# ============================================================================
# 6. NOTES FOR REAL DATASET
# ============================================================================
print("\n" + "="*80)
print("📌 IMPORTANT NOTES:")
print("="*80)
print("""
For a REAL dataset, you need:

1. VIDEO FILES:
   - Place actual video files (.mp4, .avi, .mov) in data/videos/
   - File names should match the 'id' in captions.csv (e.g., vid001.mp4)
   - Recommended: At least 100+ videos per class for good performance

2. CAPTIONS:
   - Real descriptions/titles of the videos
   - Can be extracted from video metadata or manually annotated
   - More detailed captions = better text features

3. LABELS:
   - Carefully annotated by humans
   - Binary: 0 = Safe, 1 = Adult
   - Consider inter-annotator agreement for quality

4. BALANCED DATASET:
   - Try to have roughly equal samples per class
   - If imbalanced, consider using stratified sampling or SMOTE

5. ETHICAL CONSIDERATIONS:
   - Ensure legal rights to process videos
   - Follow platform guidelines
   - Use only for research/moderation purposes
""")

print("\n✅ Preprocessing complete! Data files created in data/")
print("➡️  Next step: Run 02_feature_extraction.ipynb")
```
